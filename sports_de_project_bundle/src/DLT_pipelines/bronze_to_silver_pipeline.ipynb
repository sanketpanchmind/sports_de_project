```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import *
```

```python
table_path = 'sports_de_project_catalog.bronze'

players_df = spark.read.table(f"{table_path}.bronze_players")
matches_df = spark.read.table(f"{table_path}.bronze_matches")
teams_df = spark.read.table(f"{table_path}.bronze_teams")
most_runs_avg_rate_df = spark.read.table(f"{table_path}.bronze_teamwise_home_away")
teamwise_home_away_df = spark.read.table(f"{table_path}.bronze_most_runs_avg_rate")
```

```python
players_df = players_df.dropna();
players_df = players_df.dropDuplicates();
players_df = players_df.filter(col("Bowling_Skill") != "NULL")
players_df = players_df.filter(col("Bowling_Skill").isNotNull())
players_df = players_df.dropDuplicates(["Player_Name", "DOB"]);
display(players_df)

players_df.count()
players_df.write.format("delta").mode("overwrite").saveAsTable("sports_de_project_catalog.silver.players_silver_tbl");
```

```python
matches_df = matches_df.dropna();
matches_df = matches_df.dropDuplicates();

display(matches_df)

matches_df.write.format("delta").mode("overwrite").saveAsTable("sports_de_project_catalog.silver.matches_silver_tbl");
```

```python
teams_df = teams_df.dropna();
teams_df = teams_df.dropDuplicates();

display(teams_df);

teams_df.write.format("delta").mode("overwrite").saveAsTable("sports_de_project_catalog.silver.teams_silver_tbl");
```

```python
most_runs_avg_rate_df = most_runs_avg_rate_df\
    .withColumn("away_win_percentage", round(col("away_win_percentage"),2))\
    .withColumn("home_win_percentage", round(col("home_win_percentage"),2))

most_runs_avg_rate_df = most_runs_avg_rate_df.dropna();

most_runs_avg_rate_df = most_runs_avg_rate_df.dropDuplicates(subset=["team"]);
display(most_runs_avg_rate_df)

most_runs_avg_rate_df.write.format("delta").mode("overwrite").saveAsTable("sports_de_project_catalog.silver.most_runs_avg_rate_df_silver_tbl");
```

```python

teamwise_home_away_df = teamwise_home_away_df\
    .withColumn("average", round(col("average"),2))\
    .withColumn("strikerate", round(col("strikerate"),2))\

teamwise_home_away_df = teamwise_home_away_df.dropna();

teamwise_home_away_df = teamwise_home_away_df.dropDuplicates();

display(teamwise_home_away_df)

teamwise_home_away_df.write.format("delta").mode("overwrite").saveAsTable("sports_de_project_catalog.silver.teamwise_home_away_silver_tbl");
```
