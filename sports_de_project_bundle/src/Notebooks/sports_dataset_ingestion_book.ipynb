```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import *
import datetime, uuid
```

```python
# since there are all csv files in dataset, downloaded we use that..
players_df = spark.read.format("csv").option("header","true").option("inferSchema","true").load("/Volumes/sports_de_project_catalog/sports_schema/sports_volume/Players.csv");

players_df = players_df.withColumn("_load_timestamp",current_timestamp());

```

```python
players_df.write.format("delta").mode("append").saveAsTable("sports_de_project_catalog.bronze.bronze_players");
```

```python
matches_df = spark.read.format("csv").option("header","true").option("inferSchema","true").load("/Volumes/sports_de_project_catalog/sports_schema/sports_volume/matches.csv");

matches_df = matches_df.withColumn("_load_timestamp",current_timestamp());

matches_df.printSchema();
```

```python
teams_df = spark.read.format("csv").option("header","true").option("inferSchema","true").load("/Volumes/sports_de_project_catalog/sports_schema/sports_volume/teams.csv");

teams_df = teams_df.withColumn("_load_timestamp",current_timestamp());

teams_df.printSchema()
```

```python
most_runs_avg_rate_df = spark.read.format("csv").option("header","true").option("inferSchema","true").load("/Volumes/sports_de_project_catalog/sports_schema/sports_volume/most_runs_average_strikerate.csv");

most_runs_avg_rate_df = most_runs_avg_rate_df.withColumn("_load_timestamp",current_timestamp());

most_runs_avg_rate_df.printSchema();
```

```python
teamwise_home_away_df = spark.read.format("csv").option("header","true").option("inferSchema","true").load("/Volumes/sports_de_project_catalog/sports_schema/sports_volume/teamwise_home_and_away.csv");

teamwise_home_away_df = teamwise_home_away_df.withColumn("_load_timestamp",current_timestamp())
teamwise_home_away_df.printSchema();
```

```python
from pyspark.sql.functions import lit, col
import uuid

batch_id = str(uuid.uuid4())

# Use _metadata.file_path instead of input_file_name() for Unity Catalog
players_df = players_df.withColumn("_source_file", col("_metadata.file_path"))\
    .withColumn("_batch_id", lit(batch_id));

matches_df = matches_df.withColumn("_source_file", col("_metadata.file_path"))\
    .withColumn("_batch_id", lit(batch_id));

teams_df = teams_df.withColumn("_source_file", col("_metadata.file_path"))\
    .withColumn("_batch_id", lit(batch_id));

most_runs_avg_rate_df = most_runs_avg_rate_df.withColumn("_source_file", col("_metadata.file_path"))\
    .withColumn("_batch_id", lit(batch_id));

teamwise_home_away_df = teamwise_home_away_df.withColumn("_source_file", col("_metadata.file_path"))\
    .withColumn("_batch_id", lit(batch_id));

# Display the complete schema with all metadata columns
print("Players DataFrame Schema with all metadata:")
players_df.printSchema()
```

```python
matches_df.write.format("delta").mode("append").saveAsTable("sports_de_project_catalog.bronze.bronze_matches");

teams_df.write.format("delta").mode("append").saveAsTable("sports_de_project_catalog.bronze.bronze_teams");

most_runs_avg_rate_df.write.format("delta").mode("append").saveAsTable("sports_de_project_catalog.bronze.bronze_most_runs_avg_rate");

teamwise_home_away_df.write.format("delta").mode("append").saveAsTable("sports_de_project_catalog.bronze.bronze_teamwise_home_away")    
```
